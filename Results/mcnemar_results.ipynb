# McNemar test results

This notebook displays the newest complete McNemar tests across all ranking paradigms. The input CSV is generated by `mcnemar_all.py`.

```python
from pathlib import Path
import pandas as pd

csv_path = Path('Results/mcnemar_all.csv')
if not csv_path.exists():
    csv_path = Path('mcnemar_all.csv')
results = pd.read_csv(csv_path)
# Exclude controlled reasoning/thinking-effort ablation runs for now.
if 'source' in results.columns:
    ablation_mask = results['source'].str.contains('ablation|reasoning|thinking|effort', case=False, na=False)
    results = results.loc[~ablation_mask].copy()
results.head()
```

## Full McNemar contingency table

`b` is clean-correct/attack-wrong; `c` is clean-wrong/attack-correct. A positive degradation direction means `b > c`.

```python
columns = [
    'model', 'dataset', 'paradigm', 'attack', 'prompt', 'paired',
    'both_correct', 'clean_correct_attack_wrong',
    'clean_wrong_attack_correct', 'both_wrong',
    'p_value_exact', 'p_value_exact_log10', 'p_value_holm',
    'significant_holm_0.05', 'direction'
]
display_df = results[[c for c in columns if c in results.columns]].copy()
display_df
```

## Split tables by dataset and ranking paradigm

```python
table_columns = [
    'model', 'attack', 'prompt', 'paired', 'both_correct',
    'clean_correct_attack_wrong', 'clean_wrong_attack_correct',
    'both_wrong', 'p_value_exact_log10', 'p_value_holm',
    'significant_holm_0.05', 'direction'
]
for (dataset, paradigm), subset in results.groupby(['dataset', 'paradigm'], sort=True):
    print(f'{dataset} — {paradigm.title()}')
    display(subset[[c for c in table_columns if c in subset.columns]].reset_index(drop=True))
```

## Significant results

The log10 column remains informative when the ordinary floating-point p-value underflows to zero.

```python
summary = display_df.copy()
summary['p_value'] = summary.apply(
    lambda row: f"10^({row['p_value_exact_log10']:.2f})"
    if row.get('p_value_exact', 1) == 0
    else f"{row['p_value_exact']:.3g}",
    axis=1,
)
summary = summary.rename(columns={
    'clean_correct_attack_wrong': 'b',
    'clean_wrong_attack_correct': 'c',
    'p_value_holm': 'Holm p',
})
summary[['model', 'dataset', 'paradigm', 'attack', 'prompt', 'paired', 'b', 'c', 'p_value', 'Holm p', 'significant_holm_0.05', 'direction']]
```

## Attack versus defense significance

This is a separate paired McNemar comparison: it compares the attacked predictions from the Default and Defense runs on the same instances. The p-values above compare clean versus attacked and answer a different question.

```python
import json, math, re
from pathlib import Path
from collections import defaultdict

import sys
sys.path.insert(0, str(Path('Results')))
from mcnemar_all import _correct, _key, _metadata, _qrels, exact_mcnemar_p

def prompt_mcnemar(default_source, defense_source):
    default_path, defense_path = Path(default_source), Path(defense_source)
    meta = _metadata(defense_path)
    paradigm = meta['paradigm']
    qrels = None if paradigm == 'pointwise' else _qrels(f"msmarco-passage/trec-dl-{re.search(r'20\d\d', meta['dataset']).group()}")
    records = {}
    for path, name in ((default_path, 'default'), (defense_path, 'defense')):
        payload = json.loads(path.read_text(encoding='utf-8'))
        records[name] = {_key(r, paradigm): r for r in payload if isinstance(r, dict) and r.get('phase') == 'attacked' and _key(r, paradigm) is not None}
    a = b = c = d = 0
    for key in records['default'].keys() & records['defense'].keys():
        clean = _correct(records['default'][key], paradigm, qrels)
        defense = _correct(records['defense'][key], paradigm, qrels)
        if clean is None or defense is None: continue
        if clean and defense: a += 1
        elif clean and not defense: b += 1
        elif not clean and defense: c += 1
        else: d += 1
    return a, b, c, d, exact_mcnemar_p(b, c)

comparison_rows = []
for keys, group in results.groupby(['model', 'dataset', 'paradigm', 'attack']):
    sources = dict(zip(group['prompt'], group['source']))
    if 'Default' not in sources or 'Defense' not in sources: continue
    a, b, c, d, p = prompt_mcnemar(sources['Default'], sources['Defense'])
    comparison_rows.append(dict(zip(['model','dataset','paradigm','attack'], keys), paired=a+b+c+d, default_correct_defense_correct=a, default_correct_defense_wrong=b, default_wrong_defense_correct=c, default_wrong_defense_wrong=d, p_value_exact=p, direction='defense improves' if c > b else 'defense degrades' if b > c else 'tie'))

attack_vs_defense = pd.DataFrame(comparison_rows)
attack_vs_defense
```
